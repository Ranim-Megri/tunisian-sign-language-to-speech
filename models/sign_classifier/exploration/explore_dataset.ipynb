```python
from pathlib import Path
from collections import Counter
import zipfile

ZIP_PATH = Path.home() / "Downloads" / "tunisian-sign-language-dataset.zip"
root = Path("tunisian_sign_language_dataset") 
if not root.exists():
    if not ZIP_PATH.exists():
        raise SystemExit(f"ZIP not found: {ZIP_PATH}\nDownload it from Kaggle first.")
    with zipfile.ZipFile(ZIP_PATH) as z:
        z.extractall(root)

print("Dataset path:", root.resolve())
print("\nTop-level contents:")
for item in sorted(root.iterdir()):
    print(f"{'[DIR] ' if item.is_dir() else '[FILE]'} {item.name}")

files = [p for p in root.rglob("*") if p.is_file()]
extensions = Counter(p.suffix.lower() or "[no extension]" for p in files)

print("\nFile types:")
for ext, count in extensions.most_common():
    print(f"{ext}: {count}")

folders = [p for p in root.rglob("*") if p.is_dir()]
print("\nNumber of subfolders:", len(folders))

print("\nFirst 20 files:")
for file in files[:20]:
    print(file.relative_to(root))
```
